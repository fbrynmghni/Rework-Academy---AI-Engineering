## 3. Membuat Keputusan: if / elif / else
Program menentukan aksi berdasarkan kondisi data.

```python
stok = 3

# TODO: lengkapi if/elif/else
# stok == 0  -> 'Out of stock'
# stok < 5   -> 'Low stock'
# selain itu -> 'Available'

if stok == 0:
    status = 'Out of stock'
elif stok < 5:
    status = 'Low stock'
else:
    status = 'Available'

print(status)
```

## 8. List Operations: Menambah, Menghapus, <i>Slicing</i>
List tidak statis — kita bisa menambah item baru, membuang yang lama, dan mengambil sebagian.


```python
menu = ["Kopi Susu", "Teh Manis", "Roti Bakar"]

# TODO: tambah "Nasi Goreng" ke akhir list menggunakan .append()
menu.append("Nasi Goreng")

# TODO: buang item terakhir menggunakan .pop()
menu.pop()

# TODO: cetak 2 item pertama dari menu (slicing)
print(menu[:2])

# TODO: buat list menu_upper berisi semua nama menu dalam huruf kapital (list comprehension)
menu_upper = [item.upper() for item in menu]
# print(menu_upper)
print(menu_upper)
```

## 9. Dictionary Manipulation: Update & Iterasi
Dictionary bisa diubah setelah dibuat — tambah key baru, ubah nilai, atau ambil dengan default.


```python
produk = {"nama": "Kopi Susu", "harga": 25000, "stok": 45}

# TODO: tambah key "kategori" dengan value "Minuman"
produk["kategori"] = "Minuman"

# TODO: ubah nilai "harga" menjadi 27000
produk["harga"] = 27000

# TODO: ambil nilai "diskon" dengan default 0 (pakai .get())
diskon = produk.get("diskon", 0)
print("Diskon:", diskon)

# TODO: iterasi key-value pakai .items(), cetak "key: value" tiap baris
for key, value in produk.items():
    print(f"{key}: {value}")

print(produk)
```
